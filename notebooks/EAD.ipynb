```python
import numpy as np
import pandas as pd
```

```python

df = pd.read_csv("data/weatherHistory.csv", index=)
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df["Precip Type"] = df["Precip Type"].fillna(df["Precip Type"].mode()[0])
```

Applying imputation to column "Precip Type" as it is catagorical values to remove the missing values

```python
df.isnull().sum()
```

Handled all the missing values

```python
df["Formatted Date"] = pd.to_datetime(df["Formatted Date"], utc=True) 

df["Formatted Date"].dt.day
```

```python
df["year"] =df["Formatted Date"].dt.year
df["month"] = df["Formatted Date"].dt.month
df["date"] = df["Formatted Date"].dt.day
```

```python
df.drop("Formatted Date", inplace=True, axis=1)
df.head()
```

```python
df.info()
```

```python
categorical_col = df.select_dtypes(include=['str', 'object']).columns.to_list()
categorical_col
```

```python
for col in categorical_col:
    print(df[col].mode())
```

Categorical values do not provide any insight as it has only one values throughtout the dataset. Hence it is safe to remove these columns

```python
df.drop(categorical_col, inplace=True, axis=1)
df.head()
```

```python
df.to_csv("data/cleaned-data.csv")
```
